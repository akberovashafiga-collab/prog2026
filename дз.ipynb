```python
def generate_document(name, university, subject, teacher, date):

    text = f"""
    ЗАЯВА

    Я, {name}, студент(ка) {university},
    прошу прийняти мою роботу з дисципліни "{subject}".

    Робота подається викладачу {teacher}.

    Дата подання: {date}

    Підпис: __________
    """

    return text
```

```python
import gradio as gr

demo = gr.Interface(
    fn=generate_document,
    inputs=[
        gr.Textbox(label="ПІБ студента"),
        gr.Textbox(label="Університет"),
        gr.Textbox(label="Предмет"),
        gr.Textbox(label="Викладач"),
        gr.Textbox(label="Дата")
    ],
    outputs=gr.Textbox(label="Згенерований документ"),
    title="Генератор документа за шаблоном"
)

demo.launch()
```
